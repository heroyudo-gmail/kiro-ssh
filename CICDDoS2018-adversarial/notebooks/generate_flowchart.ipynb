# Generate Flowchart - Metodologi Penelitian
Membuat diagram alur metodologi penelitian untuk paper nids-01.tex

```python
import sys
!{sys.executable} -m pip install matplotlib -q
import matplotlib.pyplot as plt
import matplotlib.patches as mpatches
from matplotlib.patches import FancyBboxPatch, FancyArrowPatch
import numpy as np

fig, ax = plt.subplots(1, 1, figsize=(4, 7), dpi=300)
ax.set_xlim(0, 4)
ax.set_ylim(0, 10)
ax.axis('off')

# Define steps
steps = [
    ('Pra-pemrosesan Data\n(78 → 68 fitur)', '#E3F2FD', '#1565C0'),
    ('Pelatihan Baseline\n(XGBoost, 68 fitur)', '#E3F2FD', '#1565C0'),
    ('Ablation Study\n(Full, Top-15, Top-10, Top-5)', '#E3F2FD', '#1565C0'),
    ('Penentuan Sweet Spot', '#FFF3E0', '#E65100'),
    ('Evasion Attack\n(Saliency Map, ε ∈ {0.01, 0.05, 0.1})', '#E3F2FD', '#1565C0'),
    ('Adversarial Training\n(Min-Max Optimization)', '#E3F2FD', '#1565C0'),
    ('Evaluasi 2×2\n(S1, S2, S3, S4)', '#E8F5E9', '#2E7D32'),
]

# Box dimensions
box_width = 3.2
box_height = 0.9
x_center = 2.0
y_start = 9.2
y_gap = 1.3

for i, (text, bg_color, border_color) in enumerate(steps):
    y = y_start - i * y_gap
    
    # Draw box
    if i == 3:  # Diamond-like for decision
        box = FancyBboxPatch(
            (x_center - box_width/2, y - box_height/2),
            box_width, box_height,
            boxstyle='round,pad=0.05',
            facecolor=bg_color, edgecolor=border_color, linewidth=1.5
        )
    else:
        box = FancyBboxPatch(
            (x_center - box_width/2, y - box_height/2),
            box_width, box_height,
            boxstyle='round,pad=0.05',
            facecolor=bg_color, edgecolor=border_color, linewidth=1.5
        )
    ax.add_patch(box)
    
    # Add text
    ax.text(x_center, y, text, ha='center', va='center',
            fontsize=7, fontweight='bold', color='#212121')
    
    # Draw arrow to next box
    if i < len(steps) - 1:
        arrow_start_y = y - box_height/2
        arrow_end_y = (y_start - (i+1) * y_gap) + box_height/2
        ax.annotate('', xy=(x_center, arrow_end_y),
                    xytext=(x_center, arrow_start_y),
                    arrowprops=dict(arrowstyle='->', color='#616161',
                                   lw=1.5, connectionstyle='arc3'))

# Add step numbers on the left
for i in range(len(steps)):
    y = y_start - i * y_gap
    ax.text(0.15, y, f'T{i+1}', ha='center', va='center',
            fontsize=6, color='#757575', fontstyle='italic')

plt.tight_layout()
plt.savefig('../data/flowchart_metodologi.png', dpi=300, bbox_inches='tight',
            facecolor='white', edgecolor='none')
plt.show()
print('Saved: ../data/flowchart_metodologi.png')
```
